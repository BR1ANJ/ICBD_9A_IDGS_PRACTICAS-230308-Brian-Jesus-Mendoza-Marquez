```python
k = 1
m = 0

# Define the stats variable with the Pokemon statistics columns you want to plot
stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed']  # Added this line to define stats

# Lista de paletas de colores que usa cada gráfico consecutivo
palette = ['magma', 'ocean', 'vlag', 'copper', 'mako', 'winter']

# Ajustamos el lienzo panorámico grande (17 de ancho por 17 de alto)
plt.figure(figsize=(17,17))

# Extraemos la lista única de tipos para el eje Y
list_types = pokemons['Type'].unique()

# Bucle para dibujar las estadísticas una por una
for i in stats:
    plt.subplot(3, 2, k)
    k = k + 1
    
    # Agrupamos por la columna correcta y sumamos solo valores numéricos
    datos_grafico = pokemons.groupby('Type').sum(numeric_only=True)[i]
    
    # Dibujamos el gráfico de barras horizontales
    sns.barplot(x=datos_grafico, y=list_types, palette=palette[m], hue=list_types, legend=False)
    m = m + 1
    
    # Añadimos el título correspondiente a cada estadística
    plt.title(f'Total of {i}')
```

# **24. Modificación del DataSet Original (3 Pokemones Nuevos y 1 Estadística - Evolution_Level)**


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pokemons = pd.read_csv("pokemon.csv")

if 'Type 1' not in pokemons.columns:
    raise ValueError("La columna 'Type 1' no existe en el dataset.")

stats = ['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed', 'Evolve']
stats = [col for col in stats if col in pokemons.columns]

if len(stats) == 0:
    raise ValueError("No se encontraron columnas estadísticas válidas.")

palettes = ['magma', 'inferno', 'copper', 'gist_heat', 'twilight_shifted', 'bone', 'copper_r']


n_stats = len(stats)
n_cols = 2
n_rows = (n_stats + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 6 * n_rows))
axes = axes.flatten()

fig.suptitle(
    'Total de estadísticas por Tipo de Pokémon',
    fontsize=20,
    fontweight='bold'
)

grouped = pokemons.groupby('Type 1').sum(numeric_only=True)

for idx, stat in enumerate(stats):
    datos_grafico = grouped[stat].sort_values(ascending=False)

    sns.barplot(
        x=datos_grafico.values,
        y=datos_grafico.index,
        hue=datos_grafico.index,
        palette=palettes[idx % len(palettes)],
        legend=False,
        ax=axes[idx]
    )

    axes[idx].set_title(f'Total de {stat}')
    axes[idx].set_xlabel(stat)
    axes[idx].set_ylabel('Tipo')

for idx in range(len(stats), len(axes)):
    axes[idx].set_visible(False)

plt.tight_layout()
plt.subplots_adjust(top=0.95)
plt.show()
```

### Análisis de Estadísticas por Generación mediante Gráficos Interactivos Bokeh


```python
from bokeh.io import output_notebook, show
from bokeh.plotting import figure
from bokeh.layouts import gridplot

output_notebook()

# Agrupar una sola vez
gen_stats = pokemons.groupby('Generation').sum(numeric_only=True)

# Obtener generaciones automáticamente
gen = gen_stats.index.tolist()

p1 = figure(width=400, height=200, title='HP for each generation')
p1.scatter(x=gen, y=gen_stats['HP'], size=5, color='red')
p1.line(x=gen, y=gen_stats['HP'], line_width=1, color='red')

p2 = figure(width=400, height=200, title='Attack for each generation')
p2.scatter(x=gen, y=gen_stats['Attack'], size=5, color='red')
p2.line(x=gen, y=gen_stats['Attack'], line_width=1, color='red')

p3 = figure(width=400, height=200, title='Defense for each generation')
p3.scatter(x=gen, y=gen_stats['Defense'], size=5, color='red')
p3.line(x=gen, y=gen_stats['Defense'], line_width=1, color='red')

p4 = figure(width=400, height=200, title='Sp. Atk for each generation')
p4.scatter(x=gen, y=gen_stats['Sp. Atk'], size=5, color='red')
p4.line(x=gen, y=gen_stats['Sp. Atk'], line_width=1, color='red')

p5 = figure(width=400, height=200, title='Sp. Def for each generation')
p5.scatter(x=gen, y=gen_stats['Sp. Def'], size=5, color='red')
p5.line(x=gen, y=gen_stats['Sp. Def'], line_width=1, color='red')

p6 = figure(width=400, height=200, title='Speed for each generation')
p6.scatter(x=gen, y=gen_stats['Speed'], size=5, color='red')
p6.line(x=gen, y=gen_stats['Speed'], line_width=1, color='red')

grid = gridplot([p1, p2, p3, p4, p5, p6], ncols=2)

show(grid)
```

Los 3 puntos clave de este código:

- El ciclo for y subplot: Automatizan el trabajo para dibujar las 6 gráficas de stats de un solo golpe, bien ordenadas hacia abajo.

- El swarmplot: Es un "gráfico de enjambre" donde cada puntito es un Pokémon. Sirve para ver dónde se amontona la mayoría.

- El color (hue): Te delata en un segundo qué tan rotos están los Pokémon legendarios en comparación con los normales dentro de cada tipo.
