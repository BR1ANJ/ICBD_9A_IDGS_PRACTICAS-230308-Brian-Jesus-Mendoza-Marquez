# **8. Limpieza de Datos, eliminar columna inecesarias para el analisis**

Dependiendo del objetivo de analisis algunas columnas no seran necesarias en nuestro dataframe , en estte caso omitiremos los datos vacios ,
por los que eliminaremos la columna 'Type 2', y renombraremos la columba por 'Type 1' por 'Type'

```python
# Check if 'Type 2' column exists before dropping it, and handle the case where it doesn't exist
if 'Type 2' in pokemons.columns:
    pokemons = pokemons.drop(columns=['Type 2']).rename(columns={'Type 1': 'Type'})
else:
    # If 'Type 2' doesn't exist, just rename 'Type 1' to 'Type'
    pokemons = pokemons.rename(columns={'Type 1': 'Type'})

# Alternative solution: Use errors='ignore' parameter to silently skip missing columns
# pokemons = pokemons.drop(columns=['Type 2'], errors='ignore').rename(columns={'Type 1': 'Type'})
```
